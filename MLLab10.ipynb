```python
import numpy as np
import pandas as pd

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
```

Part C
Association Rules in Python


```python
rng = np.random.RandomState(42)
baskets = []
for _ in range(300):
    b = set()
    if rng.rand() < 0.70: b.add("milk")
    if rng.rand() < 0.45:
        b.add("bread")
        if rng.rand() < 0.65: b.add("butter")
    if rng.rand() < 0.30:
        b.add("coffee")
        if rng.rand() < 0.70: b.add("sugar")
    if rng.rand() < 0.15: b.add("cereal")
    if rng.rand() < 0.12: b.add("eggs")
    if rng.rand() < 0.12: b.add("juice")
    if not b: b.add("milk")
    baskets.append(sorted(b))

te = TransactionEncoder()
df = pd.DataFrame(te.fit(baskets).transform(baskets), columns=te.columns_)
print(df.shape)
print(df.mean().sort_values(ascending=False).round(3))
```

```python
frequent = apriori(
    df,
    min_support=0.1,
    use_colnames=True
)

frequent["size"] = frequent["itemsets"].apply(len)

print("Frequent itemsets:")
print(frequent)

print("\nTotal:", len(frequent))
print("\nItemsets by size:")
print(frequent["size"].value_counts().sort_index())
```

```python
result = frequent[
    frequent["itemsets"] == frozenset(["bread", "butter"])
]

print(result[["support", "itemsets"]])
```

```python
fp_items = fpgrowth(
    df,
    min_support=0.1,
    use_colnames=True
)

print("FP-Growth frequent itemsets:")
print(fp_items)

print("\nTotal FP-Growth itemsets:", len(fp_items))

# Compare Apriori and FP-Growth results
apriori_set = set(frequent["itemsets"])
fp_set = set(fp_items["itemsets"])

print("\nDo Apriori and FP-Growth match?",
      apriori_set == fp_set)
```

```python
for threshold in [0.02, 0.05, 0.1, 0.2, 0.3]:
    items = apriori(
        df,
        min_support=threshold,
        use_colnames=True
    )

    print("Minimum support:", threshold)
    print("Number of frequent itemsets:", len(items))
    print("------------------------------")
```

Task 5: Association Rules

```python
rules = association_rules(
    frequent,
    metric="confidence",
    min_threshold=0.6
)

print("Total association rules:", len(rules))

print("\nAssociation Rules:")
print(
    rules[
        ["antecedents", "consequents",
         "support", "confidence", "lift"]
    ]
)

strong_rules = rules[rules["lift"] > 1.2]

print("\nNumber of strong rules:", len(strong_rules))

print("\nTop 5 strongest rules:")
print(
    strong_rules.sort_values(
        "lift", ascending=False
    )[
        ["antecedents", "consequents",
         "support", "confidence", "lift"]
    ].head(5)
)
```

Task 6 – Threshold Comparison

```python
for threshold in [0.05, 0.1, 0.2]:

    items = apriori(
        df,
        min_support=threshold,
        use_colnames=True
    )

    rules_temp = association_rules(
        items,
        metric="confidence",
        min_threshold=0.6
    )

    strong = rules_temp[rules_temp["lift"] > 1.2]

    print("Minimum support:", threshold)
    print("Total frequent itemsets:", len(items))
    print("Total association rules:", len(rules_temp))
    print("Strong rules (lift > 1.2):", len(strong))
    print("------------------------------")
```
